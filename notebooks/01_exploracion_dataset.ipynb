## Planteamiento del problema

El dataset CIC-IDS2017 contiene 2.830.743 registros distribuidos en 15 clases. 
El tráfico benigno representa, aproximadamente, el 80 % de los registros, mientras 
que el resto corresponde a diferentes tipos de ataques.

CyberSOC AI plantea el problema principal como una **detección binaria de anomalías**:

- **Normal:** tráfico etiquetado como `BENIGN`.
- **Anómalo:** tráfico correspondiente a cualquier categoría de ataque.

Las etiquetas originales de los ataques se conservarán para realizar posteriormente 
un análisis detallado del comportamiento de los modelos frente a diferentes tipos 
de amenaza.

### Hipótesis de trabajo

Se evaluará si un modelo de Deep Learning basado en autoencoder, entrenado para 
aprender el comportamiento del tráfico benigno, permite detectar tráfico anómalo 
mediante el error de reconstrucción y obtener un rendimiento competitivo frente 
a un modelo clásico supervisado de referencia.

### Criterios de evaluación

Debido al desbalance existente entre las clases, la accuracy no se 
utilizará de forma aislada. La evaluación considerará métricas como precisión, 
recall, F1-score, ROC-AUC, tasa de falsos positivos y matriz de confusión.

También se analizará el comportamiento frente a distintos tipos de ataque cuando 
la cantidad de muestras permita obtener conclusiones significativas.

```python
for file in csv_files:
    df_temp = pd.read_csv(file)
    df_temp.columns = df_temp.columns.str.strip()

    # Solo variables numéricas
    numeric = df_temp.select_dtypes(include=[np.number])

    null_count = df_temp.isnull().sum().sum()
    inf_count = np.isinf(numeric).sum().sum()

    print(f"{file.name}")
    print(f"  Registros: {len(df_temp):,}")
    print(f"  Valores nulos: {null_count:,}")
    print(f"  Valores infinitos: {inf_count:,}")
    print()
```

```python
problem_columns = {}
total_problem_rows = 0

for file in csv_files:
    df_temp = pd.read_csv(file)
    df_temp.columns = df_temp.columns.str.strip()

    numeric = df_temp.select_dtypes(include=[np.number])

    # Filas que contienen al menos un NaN o infinito
    problem_mask = (
        df_temp.isnull().any(axis=1) |
        np.isinf(numeric).any(axis=1)
    )

    problem_rows = problem_mask.sum()
    total_problem_rows += problem_rows

    # Problemas por columna
    nulls = df_temp.isnull().sum()
    infs = pd.Series(
        np.isinf(numeric).sum(),
        index=numeric.columns
    )

    columns_with_problems = (
        nulls[nulls > 0]
        .add(infs[infs > 0], fill_value=0)
        .sort_values(ascending=False)
    )

    print(f"\n{file.name}")
    print(f"Filas con algún problema: {problem_rows:,}")
    print(columns_with_problems)

print(f"\nTOTAL filas con algún NaN o infinito: {total_problem_rows:,}")
```

### Tratamiento de valores no finitos

El análisis de calidad identificó 2.867 registros que contienen al menos un valor
nulo (`NaN`) o infinito, sobre un total de 2.830.743 registros (aproximadamente
el 0,10 % del conjunto de datos).

Los valores problemáticos se concentran en las variables `Flow Bytes/s` y
`Flow Packets/s`.

Debido a la reducida proporción de registros afectados, se ha decidido eliminar
estas observaciones durante el preprocesamiento en lugar de aplicar técnicas de
imputación. De esta forma se evita introducir valores artificiales en variables
relacionadas directamente con el comportamiento del tráfico de red.

```python
total_duplicates = 0

for file in csv_files:
    df_temp = pd.read_csv(file)
    df_temp.columns = df_temp.columns.str.strip()

    duplicates = df_temp.duplicated().sum()
    total_duplicates += duplicates

    print(f"{file.name}")
    print(f"  Filas duplicadas: {duplicates:,}")
    print()

print(f"TOTAL duplicados dentro de los ficheros: {total_duplicates:,}")
```

```python
duplicate_counts_by_label = pd.Series(dtype="int64")

for file in csv_files:
    df_temp = pd.read_csv(file)
    df_temp.columns = df_temp.columns.str.strip()

    # Seleccionamos las filas consideradas duplicadas
    duplicated_rows = df_temp[df_temp.duplicated()]

    counts = duplicated_rows["Label"].value_counts()

    duplicate_counts_by_label = duplicate_counts_by_label.add(
        counts,
        fill_value=0
    )

duplicate_counts_by_label = (
    duplicate_counts_by_label
    .astype("int64")
    .sort_values(ascending=False)
)

print("Duplicados por clase:")
print(duplicate_counts_by_label)

print(f"\nTOTAL: {duplicate_counts_by_label.sum():,}")
```

### Tratamiento de registros duplicados

El análisis ha identificado 256.479 registros duplicados dentro de los distintos
ficheros del dataset, lo que representa aproximadamente un 9,1 % de los
2.830.743 registros originales.

Los duplicados no se concentran únicamente en el tráfico benigno, sino que
también afectan a determinadas categorías de ataque,
especialmente PortScan y DoS Hulk.

Se decide eliminar los registros duplicados antes de realizar la división
entre los conjuntos de entrenamiento, validación y prueba. Esta decisión busca
reducir la redundancia del conjunto de datos y evitar que
observaciones idénticas puedan aparecer simultáneamente en entrenamiento y
prueba, lo que podría producir una estimación muy optimista del
rendimiento de los modelos.

Durante la construcción del conjunto procesado se comprobarán también posibles
duplicados existentes entre ficheros diferentes.

```python
PROCESSED_DIR = Path("../data/processed")
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print(PROCESSED_DIR.resolve())
```

```python
dataframes = []

for file in csv_files:
    print(f"Procesando: {file.name}")

    df_temp = pd.read_csv(file)

    #Limpiar nombres de columnas
    df_temp.columns = df_temp.columns.str.strip()

    #Sustituir infinitos por NaN
    df_temp.replace([np.inf, -np.inf], np.nan, inplace=True)

    #Eliminar registros con valores no válidos
    df_temp.dropna(inplace=True)

    dataframes.append(df_temp)

print("\nUniendo ficheros...")

df = pd.concat(dataframes, ignore_index=True)

print(f"Registros después de eliminar NaN/inf: {len(df):,}")
```

```python
before_duplicates = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after_duplicates = len(df)
removed_duplicates = before_duplicates - after_duplicates

print(f"Registros antes de eliminar duplicados: {before_duplicates:,}")
print(f"Registros después de eliminar duplicados: {after_duplicates:,}")
print(f"Duplicados eliminados globalmente: {removed_duplicates:,}")
```

```python
df["is_attack"] = (df["Label"] != "BENIGN").astype(int)

print(df[["Label", "is_attack"]].head())
print()
print(df["is_attack"].value_counts())
```
